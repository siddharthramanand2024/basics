```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(
    "datasets/hourly_weather_dataset.csv"
)

df.head()
```

```python
print(df.shape)
print(df.dtypes)
df.isnull().sum()
```

```python
df["date"] = pd.to_datetime(
    df["date"]
)
```

```python
hour_columns = [
    f"{hour}-hour"
    for hour in range(24)
]
df["avg_temp"] = (
    df[hour_columns]
    .mean(axis=1)
)
```

```python
df["daily_max"] = (
    df[hour_columns]
    .max(axis=1)
)
```

```python
df["daily_min"] = (
    df[hour_columns]
    .min(axis=1)
)
```

```python
df["temp_range"] = (
    df["daily_max"]
    -
    df["daily_min"]
)
```

```python
df["month"] = (
    df["date"]
    .dt.month_name()
)

df["day"] = (
    df["date"]
    .dt.day
)

df["weekday"] = (
    df["date"]
    .dt.day_name()
)
```

```python
df.describe()
```

```python
hottest=df.loc[df["avg_temp"].idxmax()]
print(hottest)

coldest=df.loc[df["avg_temp"].idxmin()]
print(coldest)
```

```python
df.groupby(
    "month"
)["avg_temp"].mean()
```

```python
long_df = pd.melt(
    df,
    id_vars=["date"],
    value_vars=hour_columns,
    var_name="hour",
    value_name="temperature",
)
```

```python
long_df.groupby(
    "hour"
)["temperature"].mean()
```
