```python
from app.services.weather_service import WeatherService

from app.services.base_weather_service import (
    BaseWeatherService,
)

from app.services.forecast_weather_service import (
    ForecastWeatherService,
)

from app.services.geo_service import GeoService
```

```python
geo_service = GeoService()

coordinates = geo_service.get_coordinates(
    "hyderabad"
)

print(
    "Coordinates:",
    coordinates,
)
```

```python

# inheritance

print(
    issubclass(
        BaseWeatherService,
        WeatherService,
    )
)

print(
    issubclass(
        ForecastWeatherService,
        WeatherService,
    )
)
```

```python

# polymorphism

current_weather_service = (
    BaseWeatherService()
)

forecast_weather_service = (
    ForecastWeatherService()
)

current_weather = (
    current_weather_service
    .get_weather_data(
        17.3850,
        78.4867,
    )
)

forecast_weather = (
    forecast_weather_service
    .get_weather_data(
        17.3850,
        78.4867,
        "2026-06-20",
        2,
    )
)

print(current_weather)

print()

print(forecast_weather)
```
