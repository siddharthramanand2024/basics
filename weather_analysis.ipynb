## 3. Data Cleaning

### Check for missing values

```python
missing = df.isnull().sum()
print("Missing values per column:")
print(missing)
print()
print(f"Total missing values: {missing.sum()}")
```

### Check for duplicates

```python
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
```

### Drop duplicates (if any)

```python
df = df.drop_duplicates()
print(f"Shape after dropping duplicates: {df.shape}")
```

### Fill missing values (demonstration)

```python
# Demonstrate fillna — fill any potential NaN in numeric columns
# with the column median
numeric_cols = df.select_dtypes(
    include=["number"]
).columns

for col in numeric_cols:
    if df[col].isnull().any():
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"Filled {col} NaN with median: {median_val}")

print("No missing values remain:", df.isnull().sum().sum() == 0)
```

### Type conversion with astype()

```python
# Convert date column to datetime
df["date"] = pd.to_datetime(df["date"])

# Convert integer-like columns
df["uv_index"] = df["uv_index"].astype(int)
df["air_quality"] = df["air_quality"].astype(int)

print("Updated dtypes:")
print(df.dtypes)
```

### Rename columns (demonstration)

```python
# Demonstrate rename — rename for clarity then revert
df_renamed = df.rename(
    columns={
        "wind_speed": "wind_speed_kmh",
        "pressure": "pressure_hpa",
        "air_quality": "aqi",
    }
)
print(df_renamed.columns.tolist())

# Keep original names for consistency
del df_renamed
```

## 4. Feature Engineering

### Average temperature per day

```python
daily_avg_temp = (
    df
    .groupby("date")["temperature"]
    .transform("mean")
)
df["avg_daily_temp"] = daily_avg_temp.round(1)

df[["date", "time", "city", "temperature", "avg_daily_temp"]].head(10)
```

### Temperature category

```python
def categorize_temperature(temp):
    """Categorize temperature into Cold, Mild, Warm, or Hot."""
    if temp < 15:
        return "Cold"
    elif temp < 25:
        return "Mild"
    elif temp < 35:
        return "Warm"
    else:
        return "Hot"


df["temp_category"] = df["temperature"].apply(
    categorize_temperature
)

print("Temperature category distribution:")
print(df["temp_category"].value_counts())
```

### Extract month and weekday

```python
df["month"] = df["date"].dt.month_name()
df["month_num"] = df["date"].dt.month
df["weekday"] = df["date"].dt.day_name()

df[["date", "month", "weekday"]].head()
```

## 12. Dynamic Query Functions

Reusable functions to query the dataset dynamically.

### Function: Get daily averages for a date

```python
def get_daily_averages(df, query_date):
    """
    Get average temperature, humidity, and rainfall for a date.

    Args:
        df (pd.DataFrame): The weather dataset.
        query_date (str): Date string in YYYY-MM-DD format.

    Returns:
        pd.Series: Average values for the date.
    """
    daily = df[
        df["date"] == pd.to_datetime(query_date)
    ]
    if daily.empty:
        print(f"No data found for {query_date}")
        return None

    averages = daily[
        ["temperature", "humidity", "rainfall"]
    ].mean()

    print(f"Daily averages for {query_date}:")
    print(f"  Avg Temperature: {averages['temperature']:.1f}°C")
    print(f"  Avg Humidity:    {averages['humidity']:.0f}%")
    print(f"  Avg Rainfall:    {averages['rainfall']:.2f} mm")
    return averages


# Example usage
get_daily_averages(df, "2025-04-10")
```

### Function: Get exact record for a date and time

```python
def get_record_by_datetime(df, query_date, query_time):
    """
    Get the exact weather record for a date and time.

    Args:
        df (pd.DataFrame): The weather dataset.
        query_date (str): Date in YYYY-MM-DD format.
        query_time (str): Time in HH:00 format.

    Returns:
        pd.DataFrame: Matching record(s).
    """
    record = df[
        (df["date"] == pd.to_datetime(query_date))
        & (df["time"] == query_time)
    ]

    if record.empty:
        print(
            f"No record found for "
            f"{query_date} at {query_time}"
        )
        return record

    print(f"Record for {query_date} at {query_time}:")
    for _, row in record.iterrows():
        print(f"  City:        {row['city']}")
        print(f"  Temperature: {row['temperature']}°C")
        print(f"  Humidity:    {row['humidity']}%")
        print(f"  Pressure:    {row['pressure']} hPa")
        print(f"  Rainfall:    {row['rainfall']} mm")
        print(f"  Wind Speed:  {row['wind_speed']} km/h")
        print(f"  AQI:         {row['air_quality']}")
    return record


# Example usage
get_record_by_datetime(df, "2025-04-10", "14:00")
```
